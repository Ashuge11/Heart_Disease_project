```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = ("https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/heart-disease/"
    "processed.cleveland.data"
)
columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs", "resteceg",
    "thalach", "exang", "oldpeak", "slope", "ca", "thal", "num"
]
df = pd.read_csv(
    url,
    names = columns,
    na_values="?"
)
```

```python
df["target"] = (df["num"] > 0).astype(int)
print(df["target"].value_counts())
print(df["target"].value_counts(normalize=True).mul(100).round(2))
sns.countplot(data=df, x="target")
plt.title("Binary Target Distribution")
plt.xlabel("Target (0 = class 0, 1 = classes 1-4)")
plt.ylabel("Count")
plt.show()
```

```python
X = df.drop('target', axis=1)
y = df['target']
num_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
cat_features = ['sex', 'cp', 'fbs', 'resteceg','exang', 'slope', 'ca', 'thal']
print("X shape: ",X.shape)
print("y shape:", y.shape)
```

```python
print(X.columns.tolist())
```

```python
X = df.drop(columns = ["target", "num"])
y = df["target"]
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("All X columns:")
print(X.columns.tolist())
print("\n Numerical features:", num_features)
print("\n Catgeorical features:", cat_features)
```

```python
print("Total input features:", len(X.columns))
print("Total listed features:", len(num_features) + len(num_features) + len(cat_features))
print("Unique listed features: ", len(set(num_features + cat_features)))
```

```python
overlap = set(num_features) & set(cat_features)
print("Overlapping features:", overlap)
print("Number of Overlapping features:", len(overlap))
```

```python
unlisted_features = set(X.columns) - set(num_features + cat_features)
extra_features = set(num_features + cat_features) - set(X.columns)
print("Unlisted features:", unlisted_features)
print("Extra features: ", extra_features)
```

```python
X[num_features].describe()
```

```python
X[num_features].hist(
    figsize = (12, 8),
    bins = 20,
    edgecolor = 'black'
)
plt.tight_layout()
plt.show()
```

```python
for col in num_features:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)

    IQR = Q3-Q1
    lower =  Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = X[(X[col] < lower) | (X[col] > upper)]
    print(col, ":", len(outliers))
```

```python
for col in  num_features:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    flagged = X.loc[
       (X[col] < lower) | (X[col] > upper), col
    ]
    print(f"\n {col} ")
    print(flagged.sort_values(). to_list())
```
